```python
import pandas as pd
import numpy as np

dates = pd.date_range(
    start='2024-01-01',
    periods=365,
    freq='D'
)

sales = np.random.randint(
    100,
    500,
    size=365
)

df = pd.DataFrame({
    'date': dates,
    'sales': sales
})

display(df)
```

```python
spark_df = spark.createDataFrame(df)

display(spark_df)
```

```python
prophet_df = df.rename(
    columns={
        'date':'ds',
        'sales':'y'
    }
)

display(prophet_df)
```

```python
from prophet import Prophet

model = Prophet()

model.fit(prophet_df)
```

```python
future = model.make_future_dataframe(
    periods=30
)

future.tail()
```

```python
forecast = model.predict(future)

forecast.head()
```

```python
fig = model.plot(forecast)
display(fig)
```

```python
fig2 = model.plot_components(
    forecast
)

display(fig2)
```

```python
train = prophet_df[:-30]
test = prophet_df[-30:]

model = Prophet()
model.fit(train)

future = model.make_future_dataframe(
    periods=30
)

predictions = model.predict(future)
```

```python
from sklearn.metrics import mean_absolute_error

actual = test['y']
predicted = predictions.tail(30)['yhat']

mae = mean_absolute_error(
    actual,
    predicted
)

print("MAE:", mae)
```
